```python
# Step 1: Sort both DataFrames by all columns
df1_sorted = df1.sort_values(by=df1.columns.tolist()).reset_index(drop=True)
df2_sorted = df2.sort_values(by=df2.columns.tolist()).reset_index(drop=True)

# Step 2: Identify differences
diff_mask = df1_sorted != df2_sorted
diff_rows = df1_sorted[diff_mask.any(axis=1)]

# Step 3: Show identifying info
print("Different rows by date and vehicle_id:")
print(diff_rows[['date', 'vehicle_id']])
```

```python
# Step 1: Sort both DataFrames by all columns and reset index
df1_sorted = df1.sort_values(by=df1.columns.tolist()).reset_index(drop=True)
df2_sorted = df2.sort_values(by=df2.columns.tolist()).reset_index(drop=True)

# Step 2: Create a boolean mask of where the values differ
diff_mask = df1_sorted != df2_sorted

# Step 3: Loop over differences to get row, column, and values
differences = []

for row_idx, row in diff_mask.iterrows():
    for col in diff_mask.columns[row]:
        if row[col]:
            differences.append({
                'row': row_idx,
                'column': col,
                'date': df1_sorted.loc[row_idx, 'date'],
                'vehicle_id': df1_sorted.loc[row_idx, 'vehicle_id'],
                'df1_value': df1_sorted.loc[row_idx, col],
                'df2_value': df2_sorted.loc[row_idx, col],
            })

# Step 4: Turn into a DataFrame for easy inspection
import pandas as pd
diff_df = pd.DataFrame(differences)
print(diff_df)
```

```python
differences = []

for row_idx in range(len(df1_sorted)):
    for col in df1_sorted.columns:
        val1 = df1_sorted.at[row_idx, col]
        val2 = df2_sorted.at[row_idx, col]

        if pd.isna(val1) and pd.isna(val2):
            continue  # both are NaN — treat as equal

        if val1 != val2:
            differences.append({
                'row': row_idx,
                'column': col,
                'date': df1_sorted.at[row_idx, 'date'],
                'vehicle_id': df1_sorted.at[row_idx, 'vehicle_id'],
                'df1_value': val1,
                'df2_value': val2,
            })

diff_df = pd.DataFrame(differences)
print(diff_df)
```
